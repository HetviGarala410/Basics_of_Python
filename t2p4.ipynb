```python
d = {
    0:"",
    1: "one",
    2: "two",
    3: "three",
    4: "four",
    5: "five",
    6: "six",
    7: "seven",
    8: "eight",
    9: "nine",
    10: "ten",
    11: "eleven",
    12: "twelve",
    13: "thirteen",
    14: "fourteen",
    15: "fifteen",
    16: "sixteen",
    17: "seventeen",
    18: "eighteen",
    19: "nineteen",
    20: "twenty",
    30:"thirty",
    40:"fourty",
    50:"fifty",
    60:"sixty",
    70:"seventy",
    80:"eighty",
    90:"ninety"
}
def ans(n):
    if(n<=20):
        return d[n]

    elif(n<100):
        f=(n//10)*10
        l=n%10
        return (d[f]+" "+d[l])

    elif(n<1000):
        f=(n//100)
        x=n%100
        return d[f]+" hundred "+ans(x)

n=int(input("Enter a number"))
z=ans(n)
print(z)
```
