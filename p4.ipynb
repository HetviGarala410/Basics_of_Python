```python
n = 0
f_h = float("-inf")
s_h = float("-inf")
f_l = float("inf")
s_l = float("inf")

while n < 5:
    a = int(input("Enter a number: "))
    n = n + 1

    if a > f_h:
        s_h = f_h
        f_h = a
    elif a > s_h and a != f_h:
        s_h = a
        
    if a < f_l:
        s_l = f_l
        f_l = a
    elif a < s_l and a != f_l:
        s_l = a

print("First Max:", f_h)
print("Second Max:", s_h)
print("First Mini:", f_l)
print("Second Mini:", s_l)
```
