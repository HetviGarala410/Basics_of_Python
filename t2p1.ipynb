```python
x="Dhruv @77dhONI"
u=0
l=0
s=0
d=0
sp=0
for i in x:
    if(i.isupper()):
        u+=1
    elif(i.islower()):
        l+=1
    elif(i.isdigit()):
        d+=1
    elif(i.isspace()):
        s+=1
    else:
        sp+=1
print(f"Upper {u} Lower {l} Digits {d} Space {s} Special {sp}")
```
