```python
p=input("Enter a password")
u=0
l=0
d=0
s=0
sp=0
if(len(p)<8):
    print("Invalid Password")
else:
    for i in p:
        if(i.isupper()):
            u=u+1
        elif(i.islower()):
            l=l+1
        elif(i.isdigit()):
            d=d+1
        elif(i.isspace()):
            s=s+1
        else:
            sp=sp+1
    if(u>0 and l>0 and d>0 and sp>0 and s==0):
        print("Valid Password")
```
