```python
f=open("t3p1.txt","r")
x=f.read()
print(x)
print(f"name {f.name}")
print(f"mode {f.mode}")
print(f"read permission {f.readable()}")
print(f"write permission {f.writable()}")
print(f"close {f.closed}")
f.close()
print(f"close {f.closed}")
```

```python
f=open("1.txt","r")
x=f.read()
f.close()
f2=open("2.txt","w+")
f2.write(x[::-1])
f2.seek(0)
y=f2.read()
f2.close()
f3=open("3.txt","w")
f3.write(y.upper())
```

```python
f=open("4.txt","r")
x=f.read()
f.close()
f2=open("ipl.txt","w")
l=x.split()
for i in l:
    if(i[0]=="d"):
        f2.write(i+"\n")
f2.close()
        
```
